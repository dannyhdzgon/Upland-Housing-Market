```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

df = pd.read_csv(
    "/content/City_zhvi_bdrmcnt_3_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv"
)

# Keep only Upland, California
upland = df[
    (df["RegionName"] == "Upland") &
    (df["State"] == "CA")
]

# Money columns
money_columns = df.columns[8:]

# Display Upland, CA
display(
    upland.style.format("${:,.2f}", subset=money_columns)
)

# Last 10 years of monthly data
months = df.columns[-120:]
values = upland[months].iloc[0]

plt.figure(figsize=(14, 7))

# Full monthly line
plt.plot(months, values)

# Put a dot every 12 months
year_positions = range(0, len(months), 12)

for i in year_positions:
    plt.scatter(months[i], values.iloc[i])

    plt.annotate(
        f"${values.iloc[i]:,.0f}",
        (months[i], values.iloc[i]),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center"
    )

# Show only yearly labels
plt.xticks(
    [months[i] for i in year_positions],
    [months[i][:4] for i in year_positions],
    rotation=45
)

plt.title("Upland, CA Home Values - Last 10 Years")
plt.xlabel("Year")
plt.ylabel("Home Value ($)")

plt.gca().yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f'${x:,.0f}')
)

plt.tight_layout()
plt.show()
```

```python
value_2016 = upland["2016-07-31"].iloc[0]
value_2026 = upland["2026-07-31"].iloc[0]

growth_rate = ((value_2026 - value_2016) / value_2016) * 100

print(f"July 2016 Home Value: ${value_2016:,.2f}")
print(f"July 2026 Home Value: ${value_2026:,.2f}")
print(f"10-Year Growth Rate: {growth_rate:.2f}%")
```

```python
# Get July home value for each year
years = range(2016, 2027)

yearly_values = []

for year in years:
    date = f"{year}-07-31"
    value = upland[date].iloc[0]
    yearly_values.append(value)

# Create table
growth_table = pd.DataFrame({
    "Year": years,
    "Home Value": yearly_values
})

# Calculate year-over-year growth rate
growth_table["Growth Rate"] = growth_table["Home Value"].pct_change() * 100

# Format the table
growth_table.style.format({
    "Home Value": "${:,.2f}",
    "Growth Rate": "{:.2f}%"
})
```

```python
cagr = ((value_2026 / value_2016) ** (1/10) - 1) * 100

print(f"10-Year CAGR: {cagr:.2f}%")
```
